```
import numpy as np
stu=['a','b','c','d']
mark=[10,20,30,40]
math=[20,40,50,80]
plt.bar(stu,mark)
plt.title('bar graph')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.xticks(rotation=45)
plt.show()
plt.close()
x=np.arange(len(stu))
plt.bar(x-0.2,mark,width=0.4)
plt.bar(x+0.2,math,width=0.4)
plt.show()
plt.close()
plt.barh(x-0.2,mark,height=0.4)
plt.barh(x+0.2,math,height=0.4)
plt.xlabel('Student')
plt.ylabel('Marks')
plt.title('group bar')
plt.xticks(x,stu)
plt.legend(['Math','Mark'])
plt.show()
plt.close()
plt.bar(stu,mark)
plt.bar(stu,math,bottom=mark)
plt.show()
plt.close()
plt.barh(stu,mark)
plt.barh(stu,math,left=mark)
plt.xlabel('Student')
plt.ylabel('Marks')
plt.legend(['mark','math'])
plt.title('stack bar')
plt.show()
```
